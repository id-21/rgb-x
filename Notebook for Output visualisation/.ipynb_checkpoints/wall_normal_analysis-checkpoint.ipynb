# Wall Segmentation + Normal Vector Analysis

This notebook helps visualize and analyze RGB->X model outputs (normal vectors) and will later integrate with SegFormer for wall segmentation.

## 1. Setup & Configuration
Configure all paths here - update these with your actual paths

```python
# Path configurations - UPDATE THESE
INPUT_IMAGES_DIR = "/path/to/input/images"  # Directory with original RGB images
RGBX_OUTPUTS_DIR = "/path/to/rgbx/outputs"  # Directory with .npy files from RGB->X model
EXTERNAL_MODULE_PATH = "/path/to/external/module"  # Path to directory containing SegFormer module
OUTPUT_DIR = "/path/to/output"  # Where to save visualizations and results

# File extension for RGB->X outputs
NORMAL_NPY_SUFFIX = "_normal.npy"  # Adjust based on your naming convention

print(f"Input images: {INPUT_IMAGES_DIR}")
print(f"RGB->X outputs: {RGBX_OUTPUTS_DIR}")
print(f"External module: {EXTERNAL_MODULE_PATH}")
print(f"Output directory: {OUTPUT_DIR}")
```

## 2. Import Dependencies

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import os
from pathlib import Path
import sys

# Add external module to path (for future SegFormer import)
if EXTERNAL_MODULE_PATH not in sys.path:
    sys.path.insert(0, EXTERNAL_MODULE_PATH)
    print(f"Added {EXTERNAL_MODULE_PATH} to Python path")

# Create output directory if it doesn't exist
os.makedirs(OUTPUT_DIR, exist_ok=True)

print("All dependencies imported successfully!")
```

## 3. Load and Investigate RGB->X Normal Vector Output
Let's explore what the numpy arrays contain

```python
# Get list of available .npy files
npy_files = sorted(Path(RGBX_OUTPUTS_DIR).glob(f"*{NORMAL_NPY_SUFFIX}"))
print(f"Found {len(npy_files)} normal vector output files")

if npy_files:
    print("\nFirst few files:")
    for f in npy_files[:5]:
        print(f"  - {f.name}")
else:
    print("WARNING: No .npy files found! Check your path and suffix.")
```

```python
# Load a sample numpy array to investigate
if npy_files:
    sample_file = npy_files[0]
    print(f"Loading: {sample_file.name}\n")
    
    normal_array = np.load(sample_file)
    
    # Investigation
    print("=" * 60)
    print("ARRAY INVESTIGATION")
    print("=" * 60)
    print(f"Shape: {normal_array.shape}")
    print(f"Data type: {normal_array.dtype}")
    print(f"\nValue range:")
    print(f"  Min: {normal_array.min():.4f}")
    print(f"  Max: {normal_array.max():.4f}")
    print(f"  Mean: {normal_array.mean():.4f}")
    print(f"  Std: {normal_array.std():.4f}")
    
    # Check if it's normalized (normal vectors should have unit length)
    if len(normal_array.shape) == 3 and normal_array.shape[2] == 3:
        # Assuming shape is (H, W, 3) for normal vectors
        norms = np.linalg.norm(normal_array, axis=2)
        print(f"\nNormal vector magnitudes:")
        print(f"  Min magnitude: {norms.min():.4f}")
        print(f"  Max magnitude: {norms.max():.4f}")
        print(f"  Mean magnitude: {norms.mean():.4f}")
        print(f"  (Should be ~1.0 if normalized)")
    
    # Sample a few pixel values
    print(f"\nSample pixel values (first 3 pixels):")
    h, w = normal_array.shape[:2]
    for i in range(min(3, h)):
        print(f"  Pixel [{i}, 0]: {normal_array[i, 0]}")
    
    print("=" * 60)
else:
    print("No files to investigate!")
```

## 4. Visualize Normal Vectors as RGB Image
Normal vectors (x, y, z) can be visualized as RGB colors

```python
def normalize_for_visualization(normals):
    """
    Convert normal vectors to RGB visualization.
    Assumes normals are in range [-1, 1] or need normalization.
    """
    # If not already in [0, 1] range, normalize from [-1, 1] to [0, 1]
    vis = normals.copy()
    
    # Map from [-1, 1] to [0, 1]
    vis = (vis + 1.0) / 2.0
    
    # Clip to ensure valid range
    vis = np.clip(vis, 0, 1)
    
    return vis

def normals_to_rgb(normals):
    """
    Convert normal vectors to RGB image for visualization.
    Returns PIL Image.
    """
    vis = normalize_for_visualization(normals)
    rgb_array = (vis * 255).astype(np.uint8)
    return Image.fromarray(rgb_array)

print("Visualization functions defined!")
```

```python
# Visualize the sample normal vector array
if npy_files:
    # Load corresponding input image if available
    image_name = sample_file.stem.replace(NORMAL_NPY_SUFFIX.replace('.npy', ''), '')
    possible_extensions = ['.jpg', '.jpeg', '.png', '.JPG', '.JPEG', '.PNG']
    
    input_image = None
    for ext in possible_extensions:
        image_path = Path(INPUT_IMAGES_DIR) / f"{image_name}{ext}"
        if image_path.exists():
            input_image = Image.open(image_path)
            print(f"Found input image: {image_path.name}")
            break
    
    # Create visualization
    normal_rgb = normals_to_rgb(normal_array)
    
    # Plot
    if input_image is not None:
        fig, axes = plt.subplots(1, 2, figsize=(16, 8))
        axes[0].imshow(input_image)
        axes[0].set_title("Original Input Image", fontsize=14, fontweight='bold')
        axes[0].axis('off')
        
        axes[1].imshow(normal_rgb)
        axes[1].set_title("Normal Vectors (as RGB)", fontsize=14, fontweight='bold')
        axes[1].axis('off')
    else:
        fig, ax = plt.subplots(1, 1, figsize=(10, 8))
        ax.imshow(normal_rgb)
        ax.set_title("Normal Vectors (as RGB)", fontsize=14, fontweight='bold')
        ax.axis('off')
        print("Note: Could not find corresponding input image")
    
    plt.tight_layout()
    plt.show()
    
    # Save visualization
    output_path = Path(OUTPUT_DIR) / f"{sample_file.stem}_visualization.png"
    normal_rgb.save(output_path)
    print(f"\nSaved visualization to: {output_path}")
else:
    print("No files to visualize!")
```

## 5. Analyze Normal Vector Distribution
Let's see what directions the normals are pointing

```python
# Analyze the distribution of normal vectors
if npy_files and normal_array is not None:
    # Flatten to (N, 3) for analysis
    h, w, c = normal_array.shape
    normals_flat = normal_array.reshape(-1, 3)
    
    # Plot histograms for each component
    fig, axes = plt.subplots(1, 3, figsize=(18, 4))
    
    components = ['X', 'Y', 'Z']
    colors = ['red', 'green', 'blue']
    
    for i, (comp, color) in enumerate(zip(components, colors)):
        axes[i].hist(normals_flat[:, i], bins=100, color=color, alpha=0.7, edgecolor='black')
        axes[i].set_title(f"{comp} Component Distribution", fontsize=12, fontweight='bold')
        axes[i].set_xlabel(f"{comp} value")
        axes[i].set_ylabel("Frequency")
        axes[i].grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    print(f"\nTotal pixels analyzed: {normals_flat.shape[0]:,}")
else:
    print("No data to analyze!")
```

## 6. Batch Processing Helper
Process multiple images at once

```python
def process_and_visualize_batch(num_images=3):
    """
    Process and visualize multiple RGB->X outputs
    """
    files_to_process = npy_files[:min(num_images, len(npy_files))]
    
    for npy_file in files_to_process:
        print(f"\nProcessing: {npy_file.name}")
        print("-" * 60)
        
        # Load normal array
        normals = np.load(npy_file)
        print(f"Shape: {normals.shape}, Range: [{normals.min():.3f}, {normals.max():.3f}]")
        
        # Convert to RGB visualization
        normal_rgb = normals_to_rgb(normals)
        
        # Try to find input image
        image_name = npy_file.stem.replace(NORMAL_NPY_SUFFIX.replace('.npy', ''), '')
        input_image = None
        
        for ext in ['.jpg', '.jpeg', '.png', '.JPG', '.JPEG', '.PNG']:
            image_path = Path(INPUT_IMAGES_DIR) / f"{image_name}{ext}"
            if image_path.exists():
                input_image = Image.open(image_path)
                break
        
        # Display
        if input_image:
            fig, axes = plt.subplots(1, 2, figsize=(14, 7))
            axes[0].imshow(input_image)
            axes[0].set_title(f"Input: {image_name}", fontsize=11)
            axes[0].axis('off')
            
            axes[1].imshow(normal_rgb)
            axes[1].set_title("Normal Vectors (RGB)", fontsize=11)
            axes[1].axis('off')
        else:
            fig, ax = plt.subplots(1, 1, figsize=(8, 7))
            ax.imshow(normal_rgb)
            ax.set_title(f"Normal Vectors: {npy_file.stem}", fontsize=11)
            ax.axis('off')
        
        plt.tight_layout()
        plt.show()
        
        # Save
        output_path = Path(OUTPUT_DIR) / f"{npy_file.stem}_vis.png"
        normal_rgb.save(output_path)
        print(f"Saved: {output_path.name}")

print("Batch processing function ready!")
print("Run: process_and_visualize_batch(num_images=5) to process multiple images")
```

## 8. Quick Reference: Color Interpretation

When viewing normal vectors as RGB:
- **Red channel (X)**: Left (-) to Right (+)
- **Green channel (Y)**: Down (-) to Up (+) 
- **Blue channel (Z)**: Away from camera (-) to Toward camera (+)

Common patterns:
- **Cyan/Blue-ish**: Surfaces facing the camera
- **Yellow**: Right-facing vertical surfaces
- **Magenta**: Left-facing vertical surfaces
- **Dark/Black**: Surfaces facing away from camera
